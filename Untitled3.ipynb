```python
import pandas as ut

df1=ut.DataFrame({
    'id': [1,2,3,4],
    'name': ['tony','strange','thor','hulk'],
    'age': [10,20,30,40]

})
df2=ut.DataFrame({
    'id': [3,4,6,7],
    'name': ['dr.doom','thanos','spiderman','andrew'],
    'age': [44,55,68,96]

})
#the first data frame is treated as left and other is treated as right
#in the left it will print all the data and print the all data and print NaN

df3=ut.merge(df1,df2, on = 'id', how = 'left')
print(df3)

```

```python
import pandas as ut

df1=ut.DataFrame({
    'id': [1,2,3,4],
    'name': ['tony','strange','thor','hulk'],
    'age': [10,20,30,40]

})
df2=ut.DataFrame({
    'id': [3,4,6,7],
    'name': ['dr.doom','thanos','spiderman','andrew'],
    'age': [44,55,68,96]

})

#only print the common data

df3=ut.merge(df1,df2, on = 'id', how = 'inner')
print(df3)

```
